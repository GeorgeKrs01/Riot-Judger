```python
# A list of all the participants puuids
participants = meta['metadata']['participants']

# Now, find where in the data our players puuid is found (Index)
player_index = participants.index(puuid)
player_index
```

```python
#Check if is the correct champion
meta['info']['participants'][player_index]['championName']
```

```python
# player_index = 1

championName = meta['info']['participants'][player_index]['championName']
print("Champion: ", championName)

game_duration = meta['info']['gameDuration']
game_duration_seconds = round(meta['info']['gameDuration'])
minutes = game_duration_seconds // 60
seconds = game_duration_seconds % 60

print(f"Game Duration {minutes}:{seconds:02} mins")

ded = meta['info']['participants'][player_index]['totalTimeSpentDead']
totalTimeSpentDead = round((meta['info']['participants'][player_index]['totalTimeSpentDead'])/60, 2)
print("DED for: ", totalTimeSpentDead , "minutes")

percentageDed = round(((ded / game_duration)*100), 2)
print(f"Percentage of the game ded {percentageDed} %")

kda = round(meta['info']['participants'][player_index]['challenges']['kda'], 2)
print("KDA: ", kda)

skillshotsDodged = meta['info']['participants'][player_index]['challenges']['skillshotsDodged']
print("Dddddodged: ", skillshotsDodged)

skillshotsHit = meta['info']['participants'][player_index]['challenges']['skillshotsHit']
print("Sniper (skillshots landed): ", skillshotsHit)

teamDamagePercentage = round(meta['info']['participants'][player_index]['challenges']['teamDamagePercentage'], 2)*100
print("I carry you (% of damage in team): ", teamDamagePercentage, "%")

longestTimeSpentLiving = round((meta['info']['participants'][player_index]['longestTimeSpentLiving'])/60, 2)
print("I shall not die (longest time alive): ", longestTimeSpentLiving, "mins")

win = meta['info']['participants'][player_index]['win']
print("WIN? ", win)
```

```python
#get the timeline data 

timelineData = requests.get(f"https://europe.api.riotgames.com/lol/match/v5/matches/{lastMatch}/timeline?api_key={API_KEY}")
timelineData = timelineData.json()
timelineData
```

```python
timelineDataInfo = timelineData['info']
timelineDataInfo.keys()
```

```python
timelineDataInfoFrames = timelineDataInfo['frames']
timelineDataInfoFrames
```

```python
type(timelineDataInfoFrames)
```

```python
import pandas as pd 

data = timelineDataInfoFrames

# Extract participantFrames -> championStats
frames = data[0]['participantFrames']
df = pd.DataFrame([stats['championStats'] for pid, stats in frames.items()])
print(df)
```

```python
from dataclasses import dataclass

@dataclass
class ChampionStats:
    abilityHaste: int
    abilityPower: int
    armor: int
    armorPen: int
    armorPenPercent: int
    attackDamage: int
    attackSpeed: int
    bonusArmorPenPercent: int
    bonusMagicPenPercent: int
    ccReduction: int
    cooldownReduction: int
    health: int
    healthMax: int
    healthRegen: int
    lifesteal: int
    magicPen: int
    magicPenPercent: int
    magicResist: int
    movementSpeed: int
    omnivamp: int
    physicalVamp: int
    power: int
    powerMax: int
    powerRegen: int
    spellVamp: int

# Convert the dictionary to an object
stats_dict = data[0]['participantFrames']['1']['championStats']
champ_stats = ChampionStats(**stats_dict)
print(champ_stats)
```
